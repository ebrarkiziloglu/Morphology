```python
import pandas as pd
from german_ud_cases_load import load_case_dictionary
import csv
```

```python
dictionary = load_case_dictionary("german_ud_cases_dictionary.csv")
```

```python
duplicate_entries = {}

for key, value in dictionary.items():
    for entry in value:
        if len(entry['forms']) > 1:
            if key not in duplicate_entries:    
                duplicate_entries[key] = []
            duplicate_entries[key].append(entry)

duplicate_entries
```

```python
duplicate_entries_dict_path = "german_ud_cases_dictionary_duplicates.csv"
with open(duplicate_entries_dict_path, 'w', encoding='utf-8', newline='') as f:
        writer = csv.writer(f)
        writer.writerow(["Lemma", "Case", "Number", "Gender", "Degree", "Upos", "Forms"])
        for lemma in sorted(duplicate_entries.keys()):
            cases = duplicate_entries[lemma]
            for case in cases:  
                key = (case['Case'], case['Number'], case['Gender'], case['Degree'], case['Upos'])
                forms = case['forms']
                forms_str = " | ".join(sorted(forms))
                writer.writerow(
                    [lemma, key[0], key[1], key[2], key[3], key[4], forms_str])
```

```python
import csv
duplicate_entries_dict_path="german_ud_cases_dictionary_duplicates.csv"
duplicate_dict = csv.reader(
    open(duplicate_entries_dict_path, 'r', encoding='utf-8'))

dict_forms = set()
# iterate over the duplicate_dict
for row in duplicate_dict:
    forms_str = set(f.strip() for f in row[6].split("|") if f.strip())
    for form in forms_str:
        dict_forms.add(form)

len(dict_forms)
```

```python
import json

# read the json file:
with open("testset/dative_accusative_pairs_group_duplicates.json",
          'r',
          encoding='utf-8') as f:
    dupl_set = json.load(f)

testset_forms = set()
for example in dupl_set:
    forms = example['gold_span_duplicate_forms']
    for form in forms:
        testset_forms.add(form)

len(testset_forms)
```
